```python
columnsToDrop = ['Unnamed: 0', 'asin','product_url', 'product_photo',
       'product_num_offers','product_availability',
       'unit_price', 'unit_count']

data.drop(columns = columnsToDrop,inplace = True)
```

```python
data.duplicated().sum()
```

```python
data['product_original_price'] = data['product_original_price'].fillna(0)
```

```python
data.dropna(inplace=True)
```

```python
data.isna().sum()
```

```python
data.info()
```

```python
data.head()
```

```python
data['product_price'] = data['product_price'].str.strip()
data['product_price'] = data['product_price'].str.replace(' ',"")
data['product_price'] = data['product_price'].str.replace('$',"")
data['product_price'] = data['product_price'].astype(float)
```
