```python
import pandas as pd
```

```python
data = pd.read_csv('Zomato.csv')
```

```python
data.info()
```

```python
data.isna().sum()
```

```python
data.drop(columns=['dish_liked'],inplace=True)
```

```python
data.isna().sum()
```

```python
data.columns = data.columns.str.strip().str.lower()
```

```python
data.head()
```

```python
data['online_order'] = data['online_order'].apply(lambda x : 1 if x=='Yes' else 0)
```

```python
data.head()
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.shape
```

```python
data['menu_item'].head()
```

```python
data = data.drop(columns = ['menu_item'])
```

```python
data = data.dropna()
```

```python
data.shape
```

```python
type(data['approx_cost(for two people)'][0])
```

```python
data['approx_cost(for two people)'] = data['approx_cost(for two people)'].str.replace(',', '', regex=False).astype(int)
```

```python
data = data.reset_index()
```
